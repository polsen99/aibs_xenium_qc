## Run batch of experiments

```python
# Fill in experiment_list with the experiment IDs (e.g., 'output-XETG...', ) as strings (in quotes) and separated by commas

experiment_list = []
```

```python
for i in experiment_list:
    try:
        exp = Experiment(i)
        exp.run_xenium_qc(run_pixel_classification=True, plot_figures=True)
        print(f'Running xenium QC for {i}: completed successfully.')
        print(' ')
    except Exception as e:
        print(f'Error running xenium QC for {i}: {e}')
```
